```python
for category, video_ids in categories.items():
    category_aggregates = {mod: {"mrr": [], "mean_rank": [], "avg_sim": [], "hit_rates": {k: [] for k in k_values}} for mod in modalities}
    
    for video_id in video_ids:
        for modality in modalities:
            metrics = evaluate_retrieval(video_id, modality=modality, k_values=k_values)
            if not metrics:
                print(f"Skipping {video_id} ({modality}): No metrics returned")
                continue
            
            with mlflow.start_run(run_name=f"{category}_{video_id}_{modality}"):
                mlflow.log_param("category", category)
                mlflow.log_param("video_id", video_id)
                mlflow.log_param("modality", modality)
                mlflow.log_param("total_queries", metrics.get("total_queries", 0))
                mlflow.log_metric("mrr", metrics["mrr"])
                mlflow.log_metric("mean_rank", metrics["mean_rank"])
                mlflow.log_metric("avg_similarity", metrics["avg_similarity"])
                for k, hr in metrics["hit_rates"].items():
                    mlflow.log_metric(f"hit_rate{k}", hr)
            
            category_aggregates[modality]["mrr"].append(metrics["mrr"])
            category_aggregates[modality]["mean_rank"].append(metrics["mean_rank"])
            category_aggregates[modality]["avg_sim"].append(metrics["avg_similarity"])
            for k in k_values:
                category_aggregates[modality]["hit_rates"][k].append(metrics["hit_rates"][k])
    
    with mlflow.start_run(run_name=f"{category}_aggregate"):
        mlflow.log_param("category", category)
        for modality in modalities:
            agg = category_aggregates[modality]
            if agg["mrr"]:
                mlflow.log_metric(f"{modality}_avg_mrr", np.mean(agg["mrr"]))
                mlflow.log_metric(f"{modality}_avg_mean_rank", np.mean(agg["mean_rank"]))
                mlflow.log_metric(f"{modality}_avg_similarity", np.mean(agg["avg_sim"]))
                for k in k_values:
                    mlflow.log_metric(f"{modality}_avg_hit_rate{k}", np.mean(agg["hit_rates"][k]))
```

```python
from google import genai
from dotenv import load_dotenv
import mlflow
import numpy as np
import os
from deepeval import evaluate
from deepeval.metrics import ContextualRelevancyMetric, ContextualPrecisionMetric, ContextualRecallMetric
from deepeval.test_case import LLMTestCase
from deepeval.models.base_model import DeepEvalBaseLLM
from evals.match_queries import evaluate_retrieval, fetch_texteval_rows, fetch_imageeval_rows, match_text_queries, match_image_queries
load_dotenv()
```

```python
from deepeval.metrics import AnswerRelevancyMetric

answer_relevancy = AnswerRelevancyMetric()
```

```python
relevancy_metric = ContextualRelevancyMetric(threshold=0.5)
precision_metric = ContextualPrecisionMetric(threshold=0.5)
recall_metric = ContextualRecallMetric(threshold=0.5)
```

```python
categories = {
    "tutorial": ["MXcsW613msA"]
}

k_values = [1, 3, 5]
modalities = ["text", "image"]

def create_deepeval_test_cases(video_id: str, modality: str):
    if modality == "text":
        rows = fetch_texteval_rows(video_id)
        match_func = match_text_queries
        context_field = "chunk"
        id_field = "chunk_id"
    
    test_cases = []
    print(f"Creating test cases for {video_id}, {modality}. Rows fetched: {len(rows)}")
    for row in rows:
        query = row["query"]
        ground_truth = row[context_field]
        results = match_func(query, video_id)
        if not results:
            print(f"No results for {video_id}, {modality}, query: {query}")
            continue
        retrieval_context = [r[context_field] for r in results[:5]]
        test_case = LLMTestCase(
            input=query,
            actual_output="\n\n".join(retrieval_context),  
            expected_output=ground_truth,
            retrieval_context=retrieval_context                   
        )
        test_cases.append(test_case)
    print(f"Test cases created: {len(test_cases)}")
    return test_cases
```

```python
category_aggregates = {cat: {mod: {"relevancy": [], "precision": [], "recall": []} for mod in modalities} for cat in categories}

for category, video_ids in categories.items():
    for video_id in video_ids:
        for modality in modalities:
            test_cases = create_deepeval_test_cases(video_id, modality)
            if not test_cases:
                print(f"No test cases for {video_id} ({modality})")
                continue
            
            results = evaluate(
                test_cases=test_cases,
                metrics=[relevancy_metric, precision_metric, recall_metric]
            )
            for test_case, scores in results:
                category_aggregates[category][modality]["relevancy"].append(
                scores.get("contextual_relevancy", 0)
                )
                category_aggregates[category][modality]["precision"].append(
                scores.get("contextual_precision", 0)
                )
                category_aggregates[category][modality]["recall"].append(
                scores.get("contextual_recall", 0)
                )
print("\n=== DeepEval Performance Analysis Across Categories and Modalities ===")
for category in categories:
    print(f"\nCategory: {category}")
    for modality in modalities:
        agg = category_aggregates[category][modality]
        if agg["relevancy"]:
            print(f"  {modality.upper()}: Avg Relevancy={np.mean(agg['relevancy']):.4f}, "
                  f"Avg Precision={np.mean(agg['precision']):.4f}, Avg Recall={np.mean(agg['recall']):.4f}")
        else:
            print(f"  {modality.upper()}: No data")

print("\nTo view DeepEval dashboard: Run `deepeval view` (opens http://localhost:8001)")
```
